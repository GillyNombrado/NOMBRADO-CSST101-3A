3. Numerical Operators


```r
x <- 10
y <- 3


add <- x + y
sub <- x - y
mul <- x * y
div <- x / y
mod <- x %% y
exp <- x ^ y

print(add)
print(sub)
print(mul)
print(div)
print(mod)
print(exp)
```

8. Vectors

```r
grades <- c(98, 89, 91, 88, 96)

grades

cat ("MEAN: " ,mean(grades))
cat ("\nMAX: ",max(grades))
cat ("\nMIN: ",min(grades))
cat ("\nLENGTH: ",length(grades))
cat ("\nMEDIAN: ",median(grades))
cat ("\nSUMMARY: ",summary(grades))
summary(grades)
```
