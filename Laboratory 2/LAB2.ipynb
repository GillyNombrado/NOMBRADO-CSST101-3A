# 2. Basic Calculation
Problem:

**A student received the following grades:**
Quiz = 85, Laboratory = 90, Project = 88, Exam = 80

**The grading weights are:** **bold text**
Quiz = 20%, Laboratory = 25%, Project = 25%, Exam = 30%

Task: Calculate the student's weighted final grade.


```r
rQuiz <- 85
rLaboratory <- 90
rProject <- 88
rExam <-80

wQuiz <- 0.20
wLaboratory <- 0.25
wProject <- 0.25
wExam <- 0.30

fQuiz <- rQuiz * wQuiz
fLaboratory <- rLaboratory * wLaboratory
fProject <- rProject * wProject
fExam <- rExam * wExam

fGrade <- fQuiz + fLaboratory + fProject + fExam
cat("The student's weighted final grade is ", fGrade)
```

# 7. Creating a Function
Instead of repeatedly writing the same code, create a function called: classify_grade()

```r
classify_grade <- function(grade) {
  if (grade >= 90 && grade <= 100) {
    return("Excellent")
  } else if (grade >= 85 && grade <= 89) {
    return("Very Good")
  } else if (grade >= 80 && grade <= 84) {
    return("Good")
  } else if (grade >= 75 && grade <= 79) {
    return("Passed")
  } else {
    return("Failed")
  }
}

print(classify_grade(75))
print(classify_grade(95))
print(classify_grade(85))
print(classify_grade(65))

```

# 10. Creating a Simple Rule-Based System
Problem:
The university wants to identify students who may need intervention.

Use the following rule:

IF attendance < 75 AND grade < 75 THEN High Risk.

Otherwise:
Low Risk.

```r
studentsInfo$Risk_Level <- ifelse(studentsInfo$attendance < 75 & studentsInfo$grade < 75, "High Risk", "Low Risk")

print(studentsInfo)
```

# 11. Apply the Rule to Multiple Students


```r
classify_risk <- function(attendance, grade) {
  if (attendance < 75 && grade < 75) {
    return("High Risk")
  } else if (attendance < 85 || grade < 80) {
    return("Moderate Risk")
  } else {
    return("Low Risk")
  }
}

studentsInfo$Risk <- mapply(classify_risk, studentsInfo$attendance, studentsInfo$grade)

print(studentsInfo)
```

# 14. Add a Recommendation
Extend your rule-based system.

For example:
* Risk - Recommendation
* High Risk - Immediate academic intervention
* Moderate Risk - Academic monitoring
* Low Risk - Continue regular monitoring

```r
get_recommendation <- function(risk_level) {
  if (risk_level == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk_level == "Moderate Risk") {
    return("Academic monitoring")
  } else if (risk_level == "Low Risk") {
    return("Continue regular monitoring")
  } else {
    return("N/A")
  }
}

studentsInfo$Recommendation <- sapply(studentsInfo$Risk, get_recommendation)

print(studentsInfo)
```
